```python
numbers = [10, 15, 22, 31, 44, 57, 63, 70, 81, 90]

odd_numbers = []

for number in numbers:
    if number % 2 != 0:
        print(number)
        odd_numbers.append(number)

print("Odd numbers list:", odd_numbers)
```
