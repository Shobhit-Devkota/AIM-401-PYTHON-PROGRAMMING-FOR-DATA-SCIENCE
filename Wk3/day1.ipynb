```python
stack = []

for i in range(5):
    number = int(input("Enter a number: "))

    # Add number to the stack
    stack.append(number)

    # Remove number if it is negative
    if number < 0:
        stack.pop()

print("Final list:", stack)

# Calculate the sum
print("Sum:", sum(stack))
```
