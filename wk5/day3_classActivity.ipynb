```python

class Khalti:
    def pay(self):
        print("Payment completed using Khalti.")


class Esewa:
    def pay(self):
        print("Payment completed using eSewa.")

class BankTransfer:
    def pay(self):
        print("Payment completed using Bank Transfer.")
```

```python
def process_payment(payment):
    payment.pay()
```

```python
khalti = Khalti()
esewa = Esewa()
bank = BankTransfer()


# 6. Process payments
process_payment(khalti)
process_payment(esewa)
process_payment(bank)
```
