```python
# Operator Overloading:

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # Addition (+)
    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    # Subtraction (-)
    def __sub__(self, other):
        return Point(self.x - other.x, self.y - other.y)

    # Multiplication (*)
    def __mul__(self, other):
        return Point(self.x * other.x, self.y * other.y)

    # Division (/)
    def __truediv__(self, other):
        return Point(self.x / other.x, self.y / other.y)

    # Floor Division (//)
    def __floordiv__(self, other):
        return Point(self.x // other.x, self.y // other.y)

    # Modulus (%)
    def __mod__(self, other):
        return Point(self.x % other.x, self.y % other.y)

    # Unary Minus (-)
    def __neg__(self):
        return Point(-self.x, -self.y)

    # Display object
    def __str__(self):
        return f"Point({self.x}, {self.y})"


# Creating objects
p1 = Point(10, 20)
p2 = Point(3, 4)

# Performing operations
print("Addition:", p1 + p2)
print("Subtraction:", p1 - p2)
print("Multiplication:", p1 * p2)
print("Division:", p1 / p2)
print("Floor Division:", p1 // p2)
print("Modulus:", p1 % p2)
print("Unary Minus:", -p1)
```

Bitwise AND using Operator Overloading
We use the special method __and__().

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __and__(self, other):
        return Point(self.x & other.x, 
                     self.y & other.y)

    def __str__(self):
        return f"Point({self.x}, {self.y})"


# Creating objects
p1 = Point(10, 20)
p2 = Point(3, 4)

# Bitwise AND
p3 = p1 & p2

print("Bitwise AND:", p3)
```
