```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def make_sound(self):
        pass
```

```python
class Dog(Animal):
    def make_sound(self):
        print("Bark")
class Cat(Animal):
    def make_sound(self):
        print("Meow")
```

```python
dog = Dog()
dog.make_sound()

cat =Cat()
cat.make_sound()


# This is abstract class so it is hidden 
# animal =Animal()
# animal.make_sound()
```
