```python
import numpy as np
```

```python
numbers = np.array(range(1,11))
```

```python
numbers
```

```python
where_result = np.where(numbers ==5)
print(where_result)
```

```python
numbers = np.array([2,8,10,5,7])
result = np.where(numbers > 5, numbers * 3, numbers)

print(result)
```

```python
arr = np.array([
    [10, -5, 20],
    [-3, 15, -8]
])

result = np.where(arr < 0, 0, arr)

print(result)
```

```python
# adding data

arr1 = np.array([1,2,3])
arr2 = np.array([4,5,6])

add = arr1 + arr2
print(add)
```

```python
combine = np.concatenate([arr1, arr2])
combine
```

```python
combine.shape
```

```python
# check matrix compatibality
if arr1.shape == arr2.shape:
    print("Arrays are compatible")
    print(arr1 + arr2)
else:
    print("Arrays are not compatible")
```

```python

original = np.array([
    [1, 2],
    [3, 4]
])

new_row = np.array([5, 6])

result = np.vstack((original, new_row))

print(result)
```

```python
# column 
new_column = np.array(([5], [6]))
add_coulum = np.hstack((original, new_column))
add_coulum
```

```python
numbers = np.array([2, 5, 10, 15, 20, 25])

result = np.clip(numbers, 5, 20)
print(result)
```

```python
# element wise addition
matrix_1 = np.array([
    [1, 2],
    [3, 4]
])

matrix_2 = np.array([
    [5, 6],
    [7, 8]
])

result = matrix_1 + matrix_2

print(result)

# element wise multiplication
mul = matrix_1 @ matrix_2
mul
```

```python
# element wise substraction
matrix_sub = matrix_1 = matrix_2
matrix_sub
```

```python
# element wise multiplication
mul = matrix_1 @ matrix_2
mul
```

```python
# element wise operation with scalars

print(matrix_1)
print("------------")
scaler =2
result_add = matrix_1 + scaler
print(result_add)
```

```python
# broadcasting 
matrix_3 = np.array([
    [1,2],
    [3,4]
    ])
array_1 = np.array([5,6])
result = matrix_3 +array_1
result
```
