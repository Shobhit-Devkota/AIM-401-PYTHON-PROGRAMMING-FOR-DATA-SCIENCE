```python
# solving linar algbra :


# Coefficient matrix
A = np.array([
    [2, 3],
    [4, 1]
])

# Constant matrix
B = np.array([8, 6])

# Solve linear equations
result = np.linalg.solve(A, B)

print("Values of x and y:", result)

print("x =", result[0])
print("y =", result[1])
```

```python
import numpy as np

# Equations:
# x + y + z = 6
# 2x + 5y + z = 4
# 2x + y + 8z = 5

A = np.array([
    [1, 1, 1],
    [2, 5, 1],
    [2, 1, 8]
])

B = np.array([6, 4, 5])

result = np.linalg.solve(A, B)

print("Solution:", result)

print("x =", result[0])
print("y =", result[1])
print("z =", result[2])
```

- X = Advertising expenditure
- Y = Sales
Given data:
Advertising (X)	Sales (Y)
1	2
2	3
3	5

We can use Linear Regression to find the slope and intercept.

```python


X = np.array([1, 2, 3])
Y = np.array([2, 3, 5])

# Calculate slope and intercept
slope, intercept = np.polyfit(X, Y, 1)

print("Slope:", slope)
print("Intercept:", intercept)

# Prediction
Y_pred = slope * X + intercept

print("Predicted Sales:", Y_pred)
```

```python
import numpy as np

X = np.array([
    [1, 1],
    [2, 1],
    [3, 1]
])

y = np.array([2, 3, 5])

# Normal Equation
beta = np.linalg.inv(X.T.dot(X)).dot(X.T).dot(y)

print("Slope (m):", beta[0])
print("Intercept (c):", beta[1])
```
