```python
class Polygon:
    def __init__(self, sides):
        self.sides = sides

    def show_sides(self):
        print(f"This polygon has {self.sides} sides.")

class Triangle(Polygon):
    def __init__(self):
        super().__init__(3)

```

```python


class Square(Polygon):
    def __init__(self):
        super().__init__(4)

class Pentagon(Polygon):
    def __init__(self):
        super().__init__(5)

class Hexagon(Polygon):
    def __init__(self):
        super().__init__(6)

```

```python
# Create objects
triangle = Triangle()
square = Square()
pentagon = Pentagon()
hexagon = Hexagon()

# Display
triangle.show_sides()
square.show_sides()
pentagon.show_sides()
hexagon.show_sides()
```
