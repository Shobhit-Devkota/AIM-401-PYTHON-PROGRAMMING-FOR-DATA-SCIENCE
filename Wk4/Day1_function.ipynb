```python
def area(length, breadth):
    a = length * breadth
    return a

print("Area: ", area(5, 6))   
```

```python
def operators():
    a = int(input("Enter the first value: "))
    b = int(input("Enter the second value: "))
    op = input("Enter the operator (+, -, *, /): ")

    if op == "+":
        result = a + b
    elif op == "-":
        result = a - b
    elif op == "*":
        result = a * b
    elif op == "/":
        if b != 0:
            result = a / b
        else:
            print("Cannot divide by zero!")
            return
    else:
        print("Invalid operator!")
        return

    print(f"Result: {result}")

operators()   
```

```python
def process_list():
    lst = []

    for i in range(5):
        value = input("Enter a number: ")

        try:
            num = int(value)
            lst.append(num)
            print(f"Integer added. Updated list: {lst}")
        except ValueError:
            try:
                num = float(value)
                if lst:
                    lst.pop()
                    print(f"Float detected, last element removed. Updated list: {lst}")
                else:
                    print(f"Float detected, but list is empty. Updated list: {lst}")
            except ValueError:
                print(f"Invalid input '{value}'. Updated list: {lst}")

    print(f"\nFinal list: {lst}")

process_list()   
```
