```python
class Calculator:

    # Class Attribute
    company = "Skill Shikshya"

    def __init__(self):
        # Instance Attributes
        self.num1 = float(input("Enter first number: "))
        self.num2 = float(input("Enter second number: "))
        self.operator = input("Enter operator (+, -, *, /): ")

    # Instance Method
    def calculate(self):

        if self.operator == "+":
            print("Result:", self.num1 + self.num2)

        elif self.operator == "-":
            print("Result:", self.num1 - self.num2)

        elif self.operator == "*":
            print("Result:", self.num1 * self.num2)

        elif self.operator == "/":
            if self.num2 != 0:
                print("Result:", self.num1 / self.num2)
            else:
                print("Cannot divide by zero!")

        else:
            print("Invalid operator!")

    # Class Method
    @classmethod
    def show_company(cls):
        print("Company:", cls.company)

    # Static Method
    @staticmethod
    def welcome():
        print("Welcome to Calculator Program")


# Static Method
Calculator.welcome()

# Class Method
Calculator.show_company()

# Create Object
calc = Calculator()

# Instance Method
calc.calculate()
```
