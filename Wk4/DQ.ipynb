```python
def factorial(n): 
    if n == 0: 
        return 1 
    return n * factorial(n - 1) 
print(factorial(5))
```

```python
def count(n): 
    if n == 0: 
        return print(n) 
    count(n - 1)
```

We can use a loop:

```python
for i in range(5, 0, -1): 
    print(i)
```

```python
numbers = [1, 2, 3, 4, 5] 
squares = [] 
for n in numbers: 
    squares.append(n * n) 
    print(squares)
```

The same operation can be written using map() and lambda:

```python
numbers = [1, 2, 3, 4, 5] 
squares = list(map(lambda n: n * n, numbers)) 
print(squares)
```

Similarly, filter() can select even numbers:

```python
even_numbers = list(filter(lambda n: n % 2 == 0, numbers)) 
print(even_numbers)
```
