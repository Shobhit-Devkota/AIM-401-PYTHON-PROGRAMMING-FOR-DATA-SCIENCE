```python
def factorial(n):
    # Base case
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)

num = int(input("Enter a number: "))
print("Factorial:", factorial(num))
```

```python
def fibonacci(n):
    # Base case
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)

n = int(input("Enter the number of terms: "))

for i in range(n):
    print(fibonacci(i), end=" ")
```

```python
# Nested function
def outer(x):
    def inner(y):
        return x + y
    return inner

add_five = outer(5)

result = add_five(6)

print(result)
```

```python
# return the function with argument:
def add(x, y):
    return x + y

def calculate(func, a, b):
    return func(a, b)

result = calculate(add, 5, 3)

print(result)
```

```python
# return a  function as a value 
def greeting(name):
    def message():
        return "Hello " + name
    return message

greet = greeting("Shobhit")

print(greet())
```

```python
# Decorator in python:
def myWrapper(func):
    def myInnerFunc():
        print("Inside Wrapper.")
        func()
    return myInnerFunc
```

```python
@ myWrapper
def myFunc():
    print("Hello")
myFunc()
```

```python
def my_decorator(func):
    def wrapper(*args, **kwargs):
        print("Function is being called")
        result = func(*args, **kwargs)
        print("Function completed")
        return result

    return wrapper


@my_decorator
def greeting(name, message="Hello"):
    print(message, name)


greeting("Shobhit", message="Good Morning")
```

```python
def count_up(n):
    i = 1

    while i <= n:
        yield i
        i += 1

gen = count_up(3)

print(next(gen))
```
