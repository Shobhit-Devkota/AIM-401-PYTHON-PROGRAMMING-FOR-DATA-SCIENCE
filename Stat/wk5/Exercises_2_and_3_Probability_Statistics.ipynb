# Exercise 2: Applying Bayes' Theorem to Fraud Detection

## Given Information

- 2% of all transactions are fraudulent.
- The detector correctly flags 90% of fraudulent transactions.
- The detector incorrectly flags 5% of legitimate transactions.


## 1. Define Probabilities



```python
p_fraud = 0.02
sensitivity = 0.90
false_positive_rate = 0.05

print("Probability of fraud:", p_fraud)
print("Sensitivity:", sensitivity)
print("False positive rate:", false_positive_rate)
```

## 2. Complement Probability




```python
p_legitimate = 1 - p_fraud

print("Probability of a legitimate transaction:", p_legitimate)
```

## 3. Total Probability

**Formula:**




```python
p_flagged = (sensitivity * p_fraud) + (false_positive_rate * p_legitimate)

print("Overall probability of a flagged transaction:", p_flagged)
```

## 4. Bayes' Theorem

**Formula:**




```python
p_fraud_given_flagged = (sensitivity * p_fraud) / p_flagged

print("Probability that a flagged transaction is fraudulent:", p_fraud_given_flagged)
print("Percentage:", p_fraud_given_flagged * 100, "%")
```

## 5. Expected Cases

**Formula:**




```python
total_flagged = 1000

expected_fraud_cases = total_flagged * p_fraud_given_flagged

print("Expected genuine fraud cases:", round(expected_fraud_cases))
```

## 6. Interpretation

The probability that a flagged transaction is actually fraudulent is not equal to the 90% sensitivity. Sensitivity measures how many fraudulent transactions the detector correctly flags. Bayes' theorem calculates how many of the flagged transactions are actually fraudulent. These are different probabilities.


# Exercise 3: Expected Value, Variance and Standard Deviation

A fair six-sided die has six possible outcomes: 1, 2, 3, 4, 5, and 6. Each outcome has an equal probability of \(1/6\).


## 1. Define Possible Outcomes

**Formula:**

\[
X = \{1, 2, 3, 4, 5, 6\}
\]


```python
outcomes = [1, 2, 3, 4, 5, 6]

print("Possible outcomes:", outcomes)
```

## 2. Calculate Probability of Each Outcome

**Formula:**

\[
P(X=x) = \frac{1}{6}
\]


```python
probability = 1 / 6

print("Probability of each outcome:", probability)
```

## 3. Calculate Expected Value E(X)

**Formula:**

\[
E(X) = \sum xP(x)
\]


```python
expected_value = sum(x * probability for x in outcomes)

print("Expected value E(X):", expected_value)
```

## 4. Calculate E(X²)

**Formula:**

\[
E(X^2) = \sum x^2P(x)
\]


```python
expected_square = sum((x ** 2) * probability for x in outcomes)

print("E(X^2):", expected_square)
```

## 5. Calculate Variance Var(X)

**Formula:**

\[
Var(X) = E(X^2) - [E(X)]^2
\]


```python
variance = expected_square - (expected_value ** 2)

print("Variance:", variance)
```

## 6. Calculate Standard Deviation SD(X)

**Formula:**

\[
SD(X) = \sqrt{Var(X)}
\]


```python
import math

standard_deviation = math.sqrt(variance)

print("Standard deviation:", standard_deviation)
```
