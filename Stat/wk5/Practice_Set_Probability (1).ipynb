# Exercise: Exploring Basic Probability 

**Scenario:** A group of 100 students was surveyed about their technical skills.

The survey produced the following results:

- 60 students know Python.
- 40 students know SQL.
- 30 students know both Python and SQL.

Let:

- **A** = A randomly selected student knows Python.
- **B** = A randomly selected student knows SQL.

### Tasks

Use Python variables and mathematical formulas to solve the following questions.

1. **Define probabilities:** Assign the given probabilities to Python variables `P_A`, `P_B`, and `P_A_and_B`.

2. **Conditional probability:** If a randomly selected student knows SQL, what is the probability that the student also knows Python? Calculate \(P(A|B)\).

3. **Union probability:** What is the probability that a randomly selected student knows Python or SQL or both? Calculate \(P(A \cup B)\).

4. **Complement probability:** What is the probability that a randomly selected student does not know Python? Calculate \(P(\neg A)\).

5. **Independence:** If knowing Python and SQL were independent events, what would be the expected probability of knowing both skills? Compare this value with the observed probability and determine whether the two events are independent.

6. **Interpretation:** Explain in your own words what the conditional probability and independence results tell us about the relationship between Python and SQL skills.


# Exercise: Applying Bayes' Theorem to Fraud Detection

Suppose an online payment platform uses an automated fraud detection system.

The following information is available:

- 2% of all transactions are fraudulent.
- The detector correctly flags 90% of fraudulent transactions.
- The detector incorrectly flags 5% of legitimate transactions.


Using the fraud detection scenario described above, answer the following questions.

**Tasks**

1. **Define probabilities:** Assign the given values to Python variables:
   - `p_fraud`
   - `sensitivity`
   - `false_positive_rate`

2. **Complement probability:** Calculate the probability that a randomly selected transaction is legitimate.

3. **Total probability:** Calculate the overall probability that a randomly selected transaction is flagged by the detector.

4. **Bayes' theorem:** Calculate the probability that a flagged transaction is actually fraudulent.

5. **Expected cases:** If the detector flags 1,000 transactions, approximately how many would you expect to be genuinely fraudulent?

6. **Interpretation:** Is the probability of a flagged transaction being fraudulent equal to the detector's 90% sensitivity? Explain your answer.


# Exercise: Expected Value, Variance and Standard Deviation

A fair six-sided die has six possible outcomes: 1, 2, 3, 4, 5 and 6.

Each outcome has an equal probability of \(1/6\).

Using Python, calculate the following:

1. Define the possible outcomes of the die.
2. Calculate the probability of each outcome.
3. Calculate the expected value \(E(X)\).
4. Calculate \(E(X^2)\).
5. Calculate the variance \(Var(X)\).
6. Calculate the standard deviation \(SD(X)\).
