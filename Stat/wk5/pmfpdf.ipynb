```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# Define values
x = np.arange(0, 11)

# Calculate PMF
y = binom.pmf(x, 10, 0.5)

# Print PMF values
print("PMF Values:", y)

# Plot PMF
plt.bar(x, y)
plt.xlabel("Number of Success")
plt.ylabel("Probability")
plt.title("Probability Mass Function (PMF)")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Define values
x = np.linspace(-4, 4, 100)

# Calculate PDF
y = norm.pdf(x, 0, 1)

# Print PDF values
print("PDF Values:", y)

# Plot PDF
plt.plot(x, y)
plt.xlabel("Values")
plt.ylabel("Density")
plt.title("Probability Density Function (PDF)")
plt.show()
```

```python

# Define values
x = np.arange(0, 11)

# Calculate CDF
y = binom.cdf(x, 10, 0.5)

# Print CDF values
print("CDF Values:", y)

# Plot CDF
plt.step(x, y, where="mid")
plt.xlabel("Number of Success")
plt.ylabel("Cumulative Probability")
plt.title("CDF - Binomial Distribution")
plt.grid()
plt.show()
```
