# Exercise 1: Exploring Basic Probability

## Given Information

- Total students = 100
- Students who know Python = 60
- Students who know SQL = 40
- Students who know both Python and SQL = 30

Let:
- **A** = Student knows Python
- **B** = Student knows SQL


## 1. Define Probabilities

**Formula:**

\[
P(A)=\frac{\text{Number of students who know Python}}{\text{Total students}}
\]

\[
P(B)=\frac{\text{Number of students who know SQL}}{\text{Total students}}
\]

\[
P(A\cap B)=\frac{\text{Number of students who know both}}{\text{Total students}}
\]


```python
total_students = 100
python_students = 60
sql_students = 40
both_students = 30

P_A = python_students / total_students
P_B = sql_students / total_students
P_A_and_B = both_students / total_students

print("P(A):", P_A)
print("P(B):", P_B)
print("P(A and B):", P_A_and_B)
```

## 2. Conditional Probability

**Formula:**

\[
P(A|B)=\frac{P(A\cap B)}{P(B)}
\]


```python
P_A_given_B = P_A_and_B / P_B

print("P(A|B):", P_A_given_B)
```

## 3. Union Probability

**Formula:**

\[
P(A\cup B)=P(A)+P(B)-P(A\cap B)
\]


```python
P_A_or_B = P_A + P_B - P_A_and_B

print("P(A or B):", P_A_or_B)
```

## 4. Complement Probability

**Formula:**

\[
P(\neg A)=1-P(A)
\]


```python
P_not_A = 1 - P_A

print("P(not A):", P_not_A)
```

## 5. Independence

**Formula:**

If A and B are independent:

\[
P(A\cap B)=P(A)\times P(B)
\]


```python
expected_both = P_A * P_B

print("Expected probability if independent:", expected_both)
print("Observed probability:", P_A_and_B)

if expected_both == P_A_and_B:
    print("Events are independent")
else:
    print("Events are dependent")
```

## 6. Interpretation

- **Conditional probability:** The probability that a student knows Python, given that the student knows SQL, is 75%.
- **Independence:** The events are dependent because the observed probability (0.30) is not equal to the expected probability if independent (0.24).
