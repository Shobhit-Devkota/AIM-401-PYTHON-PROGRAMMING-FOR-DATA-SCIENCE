# Exercise 1: Exploring Basic Probability

## Given information

A group of 100 students was asked about their technical skills.

- 60 students know Python.
- 40 students know SQL.
- 30 students know both Python and SQL.

I will use the following events:

- **A:** A student knows Python.
- **B:** A student knows SQL.

## 1. Define probabilities

First, I will store the total number of students and the number of students with each skill in variables. To find a probability, I divide the number of students in that group by the total number of students.

```python
total_students = 100

python_students = 60
sql_students = 40
both_students = 30

P_A = python_students / total_students
P_B = sql_students / total_students
P_A_and_B = both_students / total_students

print("Probability of knowing Python:", P_A)
print("Probability of knowing SQL:", P_B)
print("Probability of knowing both:", P_A_and_B)
```

The probability of knowing Python is 0.60, SQL is 0.40, and both skills is 0.30. A probability of 0.60 means 60 out of every 100 students, or 60%.

## 2. Conditional probability: P(A|B)

Conditional probability means finding the chance of one event when we already know that another event has happened.

Here, I want to find the probability that a student knows Python when I already know that the student knows SQL.

Formula:

\[
P(A|B) = \frac{P(A \cap B)}{P(B)}
\]

```python
P_A_given_B = P_A_and_B / P_B

print("Probability of knowing Python given that the student knows SQL:", P_A_given_B)
```

The answer is 0.75, or 75%. This means that among the students who know SQL, 75% also know Python.

## 3. Union probability: P(A ∪ B)

The word **or** means that a student can know Python, SQL, or both. I subtract the students who know both once because they are included in both groups.

Formula:

\[
P(A \cup B) = P(A) + P(B) - P(A \cap B)
\]

```python
P_A_or_B = P_A + P_B - P_A_and_B

print("Probability of knowing Python or SQL or both:", P_A_or_B)
```

The answer is 0.70, or 70%. This means 70% of the students know at least one of the two skills.

## 4. Complement probability: P(not A)

The complement means the probability that an event does not happen. Here, I am finding the probability that a student does not know Python.

Formula:

\[
P(\neg A) = 1 - P(A)
\]

```python
P_not_A = 1 - P_A

print("Probability of not knowing Python:", P_not_A)
```

The answer is 0.40, or 40%. Therefore, 40% of the students do not know Python.

## 5. Check whether the events are independent

Two events are independent when knowing that one event happened does not change the probability of the other event.

For independent events, the probability of both events should be equal to the product of their individual probabilities.

\[
P(A \cap B) = P(A) \times P(B)
\]

I will compare this expected value with the observed probability from the survey.

```python
expected_both_if_independent = P_A * P_B

print("Expected probability if independent:", expected_both_if_independent)
print("Observed probability of both:", P_A_and_B)

if expected_both_if_independent == P_A_and_B:
    print("The events are independent.")
else:
    print("The events are dependent.")
```

The expected probability is 0.24, but the observed probability is 0.30. These values are different, so Python and SQL knowledge are not independent in this survey.

## 6. Interpretation

From the conditional probability, I found that 75% of students who know SQL also know Python. From the independence test, I found that the two skills are dependent in this group. This means that knowing SQL is related to the chance of also knowing Python. It does not mean that learning one skill automatically causes someone to learn the other.
