```python
import pandas as pd

df = pd.read_csv(r"D:\Python_For_DataScience\Stat\assignment\nepal_student_performance.csv")

print(df.shape)         # rows, columns
df.info()               # column types and non-null counts
print(df.isna().sum())  # missing values per column
df.head()               # first 5 rows
df.describe()           # numeric summary
```

```python
df[["student_id", "exam_score"]].head(10)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.scatter(
    df["student_id"],
    df["exam_score"],
    alpha=0.6
)

plt.xlabel("Student ID")
plt.ylabel("Exam Score")
plt.title("Student ID vs Exam Score")

plt.show()
```

```python
student_ids = df["student_id"].tolist()

total = 0

for value in student_ids:
    total = total + value

mean_id = total / len(student_ids)

print("Mean student ID:", mean_id)
```

```python
import statistics

mean_id = statistics.mean(student_ids)

print("Mean student ID:", mean_id)
```

```python
std_id = statistics.stdev(student_ids)

print("Standard deviation:", std_id)
```

```python
import matplotlib.pyplot as plt

original_mean = df["student_id"].mean()

changed_ids = df["student_id"] + 10000
changed_mean = changed_ids.mean()

labels = ["Original ID", "ID + 10,000"]
means = [original_mean, changed_mean]

plt.figure(figsize=(8, 5))

plt.bar(labels, means)

plt.ylabel("Mean")
plt.title("Changing Student IDs Changes the Mean")

for i, value in enumerate(means):
    plt.text(i, value, f"{value:.2f}", ha="center", va="bottom")

plt.show()

print("Original Mean ID:", original_mean)
print("New Mean ID:", changed_mean)
```
