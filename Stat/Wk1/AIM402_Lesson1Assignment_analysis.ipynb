## Section 6: A Misleading Stored Type

`student_id` is stored as `int64` (a number), but it is really just a
**label** for each student - a nominal/identifier variable. Taking its mean
or std is meaningless. The correct summary for an ID column is a count of
**unique** values, to check for duplicates.

```python
print("Meaningless statistic (student_id treated as numeric):")
print("  Mean of student_id =", df["student_id"].mean())
print("  Std of student_id  =", df["student_id"].std())
print()
print("Correct treatment (student_id as a nominal identifier):")
print("  Unique student IDs =", df["student_id"].nunique(),
      " out of", len(df), "rows -> no duplicate records")
```

## Section 7: Frequency Table + Bar Chart (categorical variable: `stream`)

A frequency table counts how many students fall in each category.
Relative frequency = frequency / total, i.e. the proportion in each group.
The bar chart is the picture of that same table.

```python
freq = df["stream"].value_counts()
rel_freq = df["stream"].value_counts(normalize=True).round(4)
freq_table = pd.DataFrame({"Frequency": freq, "Relative Frequency": rel_freq})
freq_table
```

```python
plt.figure(figsize=(6, 4))
plt.bar(freq.index, freq.values, color="#4C72B0", edgecolor="black")
plt.title("Figure 1: Frequency of Students by Academic Stream")
plt.xlabel("Stream")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.savefig("figure1_bar_stream.png", dpi=200)
plt.show()
```

## Section 8: Histogram (numerical variable: `exam_score`)

A histogram groups a numeric variable into bins and shows how many students
fall in each bin, so you can see the overall **shape** of the distribution
(centered, skewed, spread out).

```python
df["exam_score"].describe()
```

```python
plt.figure(figsize=(6, 4))
plt.hist(df["exam_score"].dropna(), bins=20, color="#DD8452", edgecolor="black")
plt.title("Figure 2: Distribution of Exam Scores")
plt.xlabel("Exam Score")
plt.ylabel("Frequency")
plt.tight_layout()
plt.savefig("figure2_hist_examscore.png", dpi=200)
plt.show()
```
