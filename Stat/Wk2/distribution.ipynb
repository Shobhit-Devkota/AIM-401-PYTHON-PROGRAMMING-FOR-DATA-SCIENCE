```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

x = np.array([
    0.76, 1.53, 0.12, 2.44, 2.66, 3.84, 1.03, 6.31, 2.03, 4.61,
    9.53, 10.39, 3.69, 6.63, 6.4, 5.01, 2.99, 1.18, 0.35, 0.44,
    1.99, 7.04, 3.05, 7.89, 1.44, 1.11, 7.53, 10.4, 1.72, 4.96
])
```

```python
mean = np.mean(x)

print("Mean =", mean)
```

```python
median = np.median(x)

print("Median =", median)
```

```python
mode = stats.mode(x, keepdims=True)

print("Mode =", mode.mode[0])
print("Count =", mode.count[0])
```

Skewness= s3∑(x−xˉ)3/n​

```python
sum_squared = 0
sum_cubed = 0

for value in x:
    difference = value - mean

    sum_squared = sum_squared + difference ** 2
    sum_cubed = sum_cubed + difference ** 3

variance = sum_squared / len(x)
sd = variance ** 0.5

skewness = (sum_cubed / len(x)) / (sd ** 3)

print("Skewness =", skewness)
```

```python
sum_fourth = 0

for value in x:
    difference = value - mean
    sum_fourth = sum_fourth + difference ** 4

kurtosis = (sum_fourth / len(x)) / (variance ** 2)

print("Kurtosis =", kurtosis)
```

```python
print("Mean =", mean)
print("Median =", median)
print("Skewness =", skewness)
print("Kurtosis =", kurtosis)

if mean > median and skewness > 0:
    print("The data is right-skewed.")
    print("It is not symmetric like a normal distribution.")
else:
    print("The data may be approximately symmetric.")
```

```python
import matplotlib.pyplot as plt
import math

# Histogram
plt.hist(x, bins=8, density=True, alpha=0.5, edgecolor="black")

# Mean and standard deviation
sd = variance ** 0.5

# Exponential lambda
lamda = 1 / mean

# Create x values for curve
curve_x = []
normal_y = []
exponential_y = []

for i in range(0, 110):
    value = i / 10
    curve_x.append(value)

    # Normal curve
    normal = (1 / (sd * math.sqrt(2 * math.pi))) * math.exp(
        -((value - mean) ** 2) / (2 * sd ** 2)
    )
    normal_y.append(normal)

    # Exponential curve
    exponential = lamda * math.exp(-lamda * value)
    exponential_y.append(exponential)

# Plot curves
plt.plot(curve_x, normal_y, label="Normal Curve")
plt.plot(curve_x, exponential_y, label="Exponential Curve")

# Labels
plt.xlabel("Values")
plt.ylabel("Density")
plt.title("Normal vs Exponential Distribution")

plt.legend()
plt.show()
```
