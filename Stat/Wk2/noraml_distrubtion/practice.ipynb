```python
import numpy as np
import matplotlib.pyplot as plt

def normal_distribution(m):
    # Generate m random numbers
    data = np.random.normal(loc=5, scale=1.5, size=m)

    # Keep values between 1 and 10
    data = data[(data >= 1) & (data <= 10)]

    # Plot histogram
    plt.hist(data, bins=10, density=True, edgecolor="black")

    plt.xlabel("Value")
    plt.ylabel("Density")
    plt.title("Normal Distribution")
    plt.xlim(1, 10)
    plt.show()

# Call the function
normal_distribution(1000)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

def normal_distribution(m):
    # Generate m random numbers
    data = np.random.normal(loc=5, scale=1.5, size=m)

    # x values from 1 to 10
    x = np.linspace(1, 10, 500)

    # Normal distribution curve
    y = norm.pdf(x, loc=5, scale=1.5)

    # Plot curve
    plt.plot(x, y)

    plt.xlabel("Value")
    plt.ylabel("Probability Density")
    plt.title("Normal Distribution Curve")
    plt.xlim(1, 10)
    plt.grid(True)

    plt.show()

normal_distribution(1000)
```
