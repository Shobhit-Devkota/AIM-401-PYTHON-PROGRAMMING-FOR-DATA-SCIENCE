```python
# Week One: Numerical Problems

import pandas as pd
import numpy as np
```

```python
# 1. Frequency, Relative Frequency, Percentage and Cumulative Frequency

ratings = [
    'A', 'B', 'C', 'B', 'A', 'A', 'C', 'B', 'B', 'A',
    'C', 'A', 'B', 'A', 'C', 'B', 'A', 'B', 'A', 'C'
]

total = len(ratings)

frequency = pd.Series(ratings).value_counts().sort_index()

frequency_table = pd.DataFrame({
    'Frequency': frequency
})

frequency_table['Relative Frequency'] = frequency_table['Frequency'] / total
frequency_table['Percentage'] = frequency_table['Relative Frequency'] * 100
frequency_table['Cumulative Frequency'] = frequency_table['Frequency'].cumsum()

frequency_table
```

```python
# 2. Discretize a numeric variable into five levels

values = np.array([
    11.49, 9.59, 12.94, 16.57, 8.30,
    8.30, 16.74, 13.30, 6.31, 10.83,
    10.47, 10.64, 13.24, 7.09, 7.28,
    13.44, 9.99, 8.09, 9.07, 6.30
])

sorted_values = np.sort(values)

print("Sorted values:")
print(sorted_values)
```

```python
# Create 5 equal-frequency levels

levels = pd.qcut(
    values,
    q=5,
    labels=[
        'Level 1',
        'Level 2',
        'Level 3',
        'Level 4',
        'Level 5'
    ]
)

discretized_table = pd.DataFrame({
    'Original Value': values,
    'Level': levels
}).sort_values('Original Value')

discretized_table
```

```python
# Count values in each level

discretized_table['Level'].value_counts().sort_index()
```

```python
# 3. Standardize the values using Z-score

values = np.array([
    46.93, 138.07, 64.77, 51.46, 15.38,
    15.37, 11.32, 223.21, 120.23, 19.85,
    19.80, 12.39, 55.79, 86.31, 3.99,
    101.51, 30.71, 94.28, 102.39, 18.04
])

mean = np.mean(values)
std = np.std(values)

print("Mean:", round(mean, 2))
print("Standard Deviation:", round(std, 2))
```

```python
# Calculate Z-score

z_scores = (values - mean) / std

standardized_table = pd.DataFrame({
    'Original Value': values,
    'Z-score': z_scores.round(3)
})

standardized_table
```
