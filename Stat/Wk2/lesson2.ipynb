```python
# Dataset
data = [12, 36, 6, 15, 4, 22, 9, 14, 8]

# Sort data
data.sort()
n = len(data)
```

```python
print("Sorted Data:", data)
```

```python
# 1. Mean
mean = sum(data) / n
print("Mean:", mean)
```

```python
# Median
median = data[n // 2]
print("Median:", median)
```

```python
# mode
modes = []
for value in data:
    if data.count(value) > 1 and value not in modes:
        modes.append(value)

print("Mode:", modes if modes else "No mode")
```

```python
# Q1
lower_Quartile= data[: 4]
print(lower_Quartile)
# Q3
upper_Quartile =data[5:]
print(upper_Quartile)
```

```python
Q1 = (lower_Quartile[1] + lower_Quartile[2]) / 2
print("Q1:", Q1)
Q3 = (upper_Quartile[1] + upper_Quartile[2]) / 2
print("Q3:", Q3)
```

```python
# IQR
IQR = Q3 - Q1
print("IQR:", IQR)
```

```python
# Fences
lower_fence = Q1 - 1.5 * IQR
upper_fence = Q3 + 1.5 * IQR
```

```python
print(lower_fence)
print(upper_fence)
```

```python
# Range
data_range = max(data) - min(data)
print("Range:", data_range)
```

```python
# Outliers using IQR
outliers = []

for value in data:
    if value < lower_fence or value > upper_fence:
        outliers.append(value)

print("IQR Outliers:", outliers)
```

```python
# 2. Variance and Standard Deviation
squared_deviations = []

for value in data:
    squared_deviations.append((value - mean) ** 2)

SSD = sum(squared_deviations)

variance = SSD / n
SD = variance ** 0.5==

print("SSD:", SSD)
print("Population Variance:", variance)
print("Population SD:", SD)
```

```python
# Z-score for each value
print("\nZ-Scores:")

z_outliers = []

for value in data:
    z = (value - mean) / SD
    print(value, "=", round(z, 3))

    if abs(z) > 3:
        z_outliers.append(value)

print("Z-score Outliers:", z_outliers)
```

```python
# 3. Comparison
print("\nComparison:")
print("IQR Outliers:", outliers)
print("Z-score Outliers:", z_outliers)

if outliers == z_outliers:
    print("Both methods flag the same observations.")
else:
    print("The methods do not flag the same observations.")
```
