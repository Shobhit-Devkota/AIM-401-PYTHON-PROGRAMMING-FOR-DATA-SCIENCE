```python
import pandas as pd
import numpy as np

import pandas as pd
import matplotlib.pyplot as plt

# Change this path to wherever your CSV actually is on your computer
FILE_PATH = r"D:\Python_For_DataScience\Stat\assignment\nepal_student_performance.csv"

df = pd.read_csv(FILE_PATH)
df.head()
# Numerical variables that are actual measurements
numeric_variables = [
    "age",
    "study_hours",
    "sleep_hours",
    "attendance_pct",
    "family_income",
    "exam_score"
]

results = []

for column in numeric_variables:
    values = df[column].dropna()

    mean_value = values.mean()
    median_value = values.median()
    difference = mean_value - median_value
    percentage_difference = (abs(difference) / median_value) * 100

    results.append({
        "Variable": column,
        "Mean": mean_value,
        "Median": median_value,
        "Mean-Median Difference": difference,
        "Difference (%)": percentage_difference,
        "Skewness": values.skew()
    })

comparison = pd.DataFrame(results)

comparison.sort_values(
    by="Difference (%)",
    ascending=False
)
```

```python
import matplotlib.pyplot as plt

variable = "family_income"

values = df[variable].dropna()

mean_value = values.mean()
median_value = values.median()

plt.figure(figsize=(10, 5))

plt.hist(values, bins=30)

plt.axvline(
    mean_value,
    linestyle="--",
    linewidth=2,
    label=f"Mean = {mean_value:.2f}"
)

plt.axvline(
    median_value,
    linestyle="-",
    linewidth=2,
    label=f"Median = {median_value:.2f}"
)

plt.xlabel("Family Income")
plt.ylabel("Frequency")
plt.title("Distribution of Family Income with Mean and Median")

plt.legend()
plt.show()
```
