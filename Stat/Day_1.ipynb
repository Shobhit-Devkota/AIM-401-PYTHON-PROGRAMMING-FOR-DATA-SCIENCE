```python
def create_student_data():
    # Student IDs
    ids = np.array([1, 2, 3, 4, 5])

    # Minimum/pass marks
    min_marks = np.array([40, 40, 40, 40, 40])

    # Final marks
    final_marks = np.array([75, 35, 60, 90, 38])

    # Result
    result = np.where(final_marks >= min_marks, "Pass", "Fail")

    return ids, min_marks, final_marks, result
```

```python
# Call the function
ids, min_marks, final_marks, result = create_student_data()

print("ID:", ids)
print("Min Marks:", min_marks)
print("Final Marks:", final_marks)
print("Result:", result)
```

```python
import numpy as np

def create_student():
    
    # Same random data every time (optional)
    np.random.seed(42)
    
    n = 50
    
    # Student IDs from 1 to 50
    student_id = np.arange(1, 51)
    
    # Random student names
    names = np.random.choice([
        "Shobhit", "Sanjiv", "aashutoh", "ashok",
        "Ram", "Hari", "Sita", "Gita",
        "Ramesh", "Bikash"
    ], n)
    
    # Random age
    age = np.random.randint(18, 26, n)
    
    # Random study hours
    study_hours = np.random.randint(1, 11, n)
    
    # Random midterm score
    midterm_score = np.random.randint(0, 51, n)
    
    # Random final score
    final_score = np.random.randint(0, 51, n)
    
    # Random attendance percentage
    attendance = np.random.randint(50, 101, n)
    
    # Random campus
    campus = np.random.choice(
        ["Kathmandu", "Bhaktapur", "Lalitpur"],
        n
    )
    
    return {
        "Student ID": student_id,
        "Name": names,
        "Age": age,
        "Study Hours": study_hours,
        "Midterm Score": midterm_score,
        "Final Score": final_score,
        "Attendance": attendance,
        "Campus": campus
    }


```

```python
# Create dataset
data = create_student()

# Print data
for key, value in data.items():
    print(key, ":", value)
```

```python
import pandas as pd

df = pd.DataFrame(data)

df.head()
```
