```python
import pandas as pd

data = {
    "English": [56, 75, 45, 71, 62, 64, 58, 80, 76, 61],
    "Maths":   [66, 70, 40, 60, 65, 56, 59, 77, 67, 63]
}

df = pd.DataFrame(data)

print(df)
```

```python
english_sorted = sorted(df["English"])
maths_sorted = sorted(df["Maths"])

english_rank = []
maths_rank = []

for value in df["English"]:
    rank = english_sorted.index(value) + 1
    english_rank.append(rank)

for value in df["Maths"]:
    rank = maths_sorted.index(value) + 1
    maths_rank.append(rank)

df["English_Rank"] = english_rank
df["Maths_Rank"] = maths_rank

print(df)
```

```python
d = []

for i in range(len(df)):
    difference = df["English_Rank"][i] - df["Maths_Rank"][i]
    d.append(difference)

df["d"] = d

print(df)
```

```python
df["Rank_English"] = df["English"].rank()
df["Rank_Maths"] = df["Maths"].rank()
```

```python
df["d"] = df["Rank_English"] - df["Rank_Maths"]
```

```python
df["d2"] = df["d"] ** 2
```

```python
print(df)
```

##  Hypothesis testing

- **H₀:** There is no relationship between English and Maths marks in the population (correlation = 0).
- **H₁:** There is a relationship between English and Maths marks (correlation ≠ 0), two-tailed.
- **Significance level:** α = 0.05. Critical t for df = 8 is 2.306.

```python
n = len(df)

sum_d2 = 0

for value in df["d2"]:
    sum_d2 += value

rs = 1 - (6 * sum_d2) / (n * (n**2 - 1))

print("Sum of d2:", sum_d2)
print("Spearman Rank Correlation:", rs)
```

```python
from scipy.stats import pearsonr

r, p = pearsonr(df["English"], df["Maths"])

print("Pearson Correlation:", r)
print("P-value:", p)
```

```python
if p < 0.05:
    print("Reject H0: The correlation is statistically significant.")
else:
    print("Fail to reject H0: The correlation is not statistically significant.")
```

```python
from scipy.stats import pearsonr

r, p = pearsonr(df["English"], df["Maths"])

n = len(df)

t = r * ((n - 2) / (1 - r**2))**0.5

print("Pearson Correlation:", r)
print("t-value:", t)
print("p-value:", p)

if p < 0.05:
    print("Reject H0: The correlation is statistically significant.")
else:
    print("Fail to reject H0: The correlation is not statistically significant.")
```
