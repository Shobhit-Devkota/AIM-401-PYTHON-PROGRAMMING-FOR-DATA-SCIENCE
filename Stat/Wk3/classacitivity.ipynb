```python
# Data
x = [1, 2, 3, 4, 5, 6]
y = [3, 5, 4, 8, 10, 12]

# Given means
x_mean = 3.5
y_mean = 7

# Number of observations
n = len(x)

print("x =", x)
print("y =", y)
print("x mean =", x_mean)
print("y mean =", y_mean)
print("n =", n)
```

2. Calculate Sxy, Sxx and Syy

First, we calculate the differences from the mean.

```python
# Empty lists to store calculations
x_difference = []
y_difference = []

# Calculate x - x_mean and y - y_mean
for i in range(n):
    x_diff = x[i] - x_mean
    y_diff = y[i] - y_mean
    
    x_difference.append(x_diff)
    y_difference.append(y_diff)

print("x - x_mean:", x_difference)
print("y - y_mean:", y_difference)
```

Now calculate Sxy, Sxx, and Syy.

```python
# Calculate Sxy, Sxx and Syy
Sxy = 0
Sxx = 0
Syy = 0

for i in range(n):
    Sxy = Sxy + (x[i] - x_mean) * (y[i] - y_mean)
    Sxx = Sxx + (x[i] - x_mean) ** 2
    Syy = Syy + (y[i] - y_mean) ** 2

print("Sxy =", Sxy)
print("Sxx =", Sxx)
print("Syy =", Syy)
```

3. Calculate Pearson correlation coefficient

Formula:

$$ r = \frac{S_{xy}}{\sqrt{S_{xx}S_{yy}}} $$

For beginner Python, we can use math.sqrt().

```python
import math

# Calculate Pearson correlation coefficient
r = Sxy / math.sqrt(Sxx * Syy)

print("Pearson correlation coefficient (r) =", r)
```

9. Final complete code

If you want everything in one notebook cell, use this:

```python
import math

# --------------------------------
# Data
# --------------------------------

x = [1, 2, 3, 4, 5, 6]
y = [3, 5, 4, 8, 10, 12]

# Given means
x_mean = 3.5
y_mean = 7

# Number of observations
n = len(x)

print("Number of observations:", n)
print("x mean:", x_mean)
print("y mean:", y_mean)


# --------------------------------
# Calculate Sxy, Sxx and Syy
# --------------------------------

Sxy = 0
Sxx = 0
Syy = 0

for i in range(n):
    
    Sxy = Sxy + (x[i] - x_mean) * (y[i] - y_mean)
    
    Sxx = Sxx + (x[i] - x_mean) ** 2
    
    Syy = Syy + (y[i] - y_mean) ** 2

print("\nSxy =", Sxy)
print("Sxx =", Sxx)
print("Syy =", Syy)


# --------------------------------
# Pearson correlation coefficient
# --------------------------------

r = Sxy / math.sqrt(Sxx * Syy)

print("\nPearson correlation coefficient (r) =", r)


# --------------------------------
# Regression line
# --------------------------------

# Slope
b = Sxy / Sxx

# Intercept
a = y_mean - (b * x_mean)

print("\nSlope (b) =", b)
print("Intercept (a) =", a)

print("Regression line:")
print("y_hat =", a, "+", b, "* x")


# --------------------------------
# R squared
# --------------------------------

R_squared = r ** 2

print("\nR squared =", R_squared)
print("R squared percentage =", R_squared * 100, "%")


# --------------------------------
# Adjusted R squared
# --------------------------------

k = 1

adjusted_R_squared = 1 - (1 - R_squared) * (n - 1) / (n - k - 1)

print("\nAdjusted R squared =", adjusted_R_squared)
print("Adjusted R squared percentage =", adjusted_R_squared * 100, "%")


# --------------------------------
# Predicted values and residuals
# --------------------------------

predicted_y = []
residuals = []

for i in range(n):
    
    # Calculate predicted y
    y_hat = a + b * x[i]
    
    # Calculate residual
    e = y[i] - y_hat
    
    predicted_y.append(y_hat)
    residuals.append(e)


# Print table
print("\nPredicted values and residuals")
print("-----------------------------------------------")
print("x\tActual y\tPredicted y\tResidual")

for i in range(n):
    print(
        x[i],
        "\t",
        y[i],
        "\t\t",
        round(predicted_y[i], 4),
        "\t\t",
        round(residuals[i], 4)
    )


# --------------------------------
# Sum of residuals
# --------------------------------

sum_residuals = 0

for e in residuals:
    sum_residuals = sum_residuals + e

print("\nSum of residuals =", sum_residuals)
```
